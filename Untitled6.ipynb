# 1. Investigar métodos y estrategias para el Análisis Exploratorio de Datos (EDA) usando el paquete de Pandas

## Métodos principales de Pandas para EDA

head()
Muestra las primeras filas del conjunto de datos.

tail()
Muestra las últimas filas del dataset.

info()
Presenta información general como cantidad de filas, columnas y tipos de datos.

describe()
Genera estadísticas descriptivas como media, mediana, desviación estándar, mínimos y máximos.

shape
Indica la dimensión del dataset (filas y columnas).

columns
Muestra los nombres de las columnas.

isnull() y sum()
Permiten identificar valores nulos o faltantes.

value_counts()
Cuenta la frecuencia de valores en una columna.

corr()
Calcula la correlación entre variables numéricas.

## Estrategias comunes en EDA



Limpieza de datos
Eliminación de datos duplicados, corrección de errores y tratamiento de valores nulos.

Análisis estadístico
Obtención de medidas de tendencia central y dispersión.

Visualización de datos
Uso de gráficos para detectar patrones, relaciones y anomalías.

Detección de outliers
Identificación de valores atípicos que puedan afectar el análisis.

Transformación de datos
Conversión de tipos de datos o normalización de variables.

# 2. Investigar sobre paquetes de Python para generar gráficos

En Python existen múltiples bibliotecas para la visualización de datos. Estas herramientas permiten representar información mediante gráficos estadísticos e interactivos.

##Principales paquetes para generar gráficos

**Matplotlib**

Es la biblioteca más utilizada para crear gráficos básicos y avanzados en Python.

Características:

Gráficos de líneas
Barras
Histogramas
Dispersión
Personalización completa

**Seaborn**

Biblioteca basada en Matplotlib enfocada en visualizaciones estadísticas más elegantes y simples.

Características:

Heatmaps
Boxplots
Pairplots
Histogramas avanzados

**Plotly**

Permite crear gráficos interactivos y dinámicos para aplicaciones web y dashboards.

Características:

Interactividad
Zoom dinámico
Dashboards
Compatibilidad con web

**Pandas**

Además del análisis de datos, Pandas incluye funciones básicas para generar gráficos directamente desde DataFrames.
