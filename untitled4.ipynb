##Leer archivos CSV y TXT usando context managers


Primero debes de correr los archivos de w para crear y despues ejecutar el de read para leerlo, lo puse en orden invertido porque para leer archivos se deben de crear o deben de existir antes si no esto da error

```python
with open("archivo.txt", "w", encoding="utf-8") as f:
    f.write("Hola mundo\n")
    f.write("Otra línea")
```

```python
import csv

datos = [
    ["Nombre", "Edad"],
    ["Juan", 25],
    ["Ana", 30]
]

with open("archivo.csv", "w", newline='', encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(datos)
```

```python
with open("archivo.txt", "r", encoding="utf-8") as f:
    contenido = f.read()
    print(contenido)
```

```python
import csv

with open("archivo.csv", "r", newline='', encoding="utf-8") as f:
    reader = csv.reader(f)
    for fila in reader:
        print(fila)
```

El de Lectura (R) nos sirve para leer datos desde archivos, bases de datos, APIs
El de Escritura (W) bis suve Guardar datos procesados en destino final (DB, CSV, etc.) este se debe de correr antes del read porque para leer algo debe de existir algo
se puede decir
que R → inicio del pipeline
W → final y aveces enmedio

##Leer un archivo Pickle
##Escribir un archivo Pickle



al igual que en el primero aunque las intrucciones digan leer y luego escribir los codigos los puse en un orden distinto para que no de error

```python
import pickle

datos = {"nombre": "Juan", "edad": 25}

with open("archivo.pkl", "wb") as f:
    pickle.dump(datos, f)
```

```python
import pickle

with open("archivo.pkl", "rb") as f:
    datos = pickle.load(f)

print(datos)
```

En si Pickle puede guardar prácticamente cualquier objeto de Python Tiene que ser de pyton como los cuales pueden ser
Listas
