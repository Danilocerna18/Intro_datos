#context managers

Los gestores de contexto o context managers permiten asignar o liberar recursos de una forma expresa.

https://python-intermedio.readthedocs.io/es/latest/context_managers.html#:~:text=Los%20gestores%20de%20contexto%20o,te%20permiten%20hacer%20precisamente%20esto.

Los gestores de contexto permiten al usuario de Python adquirir, asignar y liberar recursos de forma segura y precisa , a menudo mediante mecanismos automáticos de configuración y desmontaje. Si bien existen numerosas maneras de hacerlo en Python, lo más común es utilizar la palabra clave "with".

https://medium.com/@jkarma0920/python-with-context-managers-53fb10dae076

Context Managers: estructuras que permiten gestionar recursos (como archivos) de manera segura y automática. Los context managers permiten manejar recursos automáticamente usando la palabra clave with.
DataFrames: estructuras de datos tabulares ampliamente utilizadas en análisis de datos.
Archivos Pickle (de esto hablaremos a en el futuro): método para serializar y guardar objetos de Python en archivos.

```python
# Abrir y leerlo
with open("archivo.txt", "r") as archivo:
    contenido = archivo.read()
    print(contenido)
```

open() abre el archivo
with asegura que el archivo se cierre automáticamente
No necesitas usar archivo.close()
algo muy importante es que siempre debes de poner si es r o w debido a que tienen funciones distintas

```python
# Escritura de archivo
with open("archivo.txt", "w") as archivo:
    archivo.write("Hola, este es un ejemplo de context manager")

# Lectura de archivo
with open("archivo.txt", "r") as archivo:
    contenido = archivo.read()
    print(contenido)
```

```python
import pandas as pd

# Crear DataFrame
data = {
    "Nombre": ["Juan", "Ana", "Luis"],
    "Edad": [23, 25, 22]
}

df = pd.DataFrame(data)

# Operaciones básicas
print(df)
print(df.head())
print(df["Edad"].mean())
```

#Archivos Pickle

Pickle es un módulo disponible en la biblioteca estándar de Python que permite transformar objetos en una secuencias de bytes, que podemos luego reconstruir en objetos en memoria

https://www.arsys.es/blog/python-pickle-que-es-y-por-que-utlizar-esta-serializacion#:~:text=Pickle%20es%20un%20m%C3%B3dulo%20disponible,reconstruir%20en%20objetos%20en%20memoria.

Pickle ofrece flexibilidad al deserializar objetos . Puedes guardar fácilmente diferentes variables en un archivo Pickle y cargarlas de nuevo en una sesión de Python diferente, recuperando tus datos exactamente como estaban sin tener que modificar tu código.

https://www.datacamp.com/tutorial/pickle-python-tutorial

Pickle permite guardar objetos de Python en archivos binarios.

Para qué sirve:
Guardar modelos de machine learning
Persistir datos sin perder estructura
Transferir objetos entre programas

```python
import pickle

# Datos a guardar
datos = {"nombre": "Juan", "edad": 23}

# Guardar (serializar)
with open("datos.pkl", "wb") as archivo:
    pickle.dump(datos, archivo)

# Cargar (deserializar)
with open("datos.pkl", "rb") as archivo:
    datos_cargados = pickle.load(archivo)

print(datos_cargados)
```

Se utiliza pickle.dump() para guardar datos.
Se utiliza pickle.load() para cargar datos.
Se emplea with para manejar correctamente el archivo esto lo vimos en el punto uno

La principal ventaja es que permite guardar y recuperar estructuras de datos exactamente como estaban, sin necesidad de reconstruirlas manualmente.
