```python
import pandas as pd
import numpy as np


np.random.seed(42)

data = {
    "id": range(1, 101),
    "edad": np.random.randint(18, 60, 100),
    "ingreso": np.random.randint(2000, 10000, 100),
    "genero": np.random.choice(["Masculino", "Femenino"], 100),
    "ciudad": np.random.choice(["Guatemala", "Antigua", "Mixco"], 100),
    "compras": np.random.randint(1, 20, 100)
}

df = pd.DataFrame(data)

print("Primeras filas:")
print(df.head())


print("\nTipos de datos:")
print(df.dtypes)

print("\nInfo del dataset:")
print(df.info())


print("\nEstadísticas:")
print(df.describe())


print("\nFrecuencia género:")
print(df["genero"].value_counts())

print("\nFrecuencia ciudad:")
print(df["ciudad"].value_counts())
```

## ¿Cómo se relaciona este análisis con las metodologías vistas en clase (DLM, DL, CRISP-DM)?


**El EDA forma parte de la etapa de Data Understanding en CRISP-DM,en donde se va  a analizar la estructura, calidad y distribución de los datos. En metodologías como DLM y DL, este análisis es muy importante antes de entrenar modelos, ya que permite identificar errores, valores atípicos y relaciones entre variables. En resumen, el EDA nos ayuda a tomar mejores decisiones antes de construir el modelo a utilizar o los pipelines.**

## ¿Cómo se podría identificar el tipo de gráfico ideal para presentar información de un dataset?


En si para poder elegir el tipo de gráfico depende del tipo de datos que estemos trabajando asi como lo que querramos hacer por ejemplo:

Scatter plot: para ver relación entre dos variables numéricas
Gráfico de barras: para comparar categorías y ver facilmente cual es mas alta
Histograma: para ver la distribución de una variable

al elegir un buen grafico se puede presentar de mejor manera la informacion

```python

plt.figure()
plt.hist(df["ingreso"], bins=10)
plt.title("Distribución de ingresos")
plt.show()

plt.figure()
plt.scatter(df["edad"], df["compras"])
plt.title("Edad vs Compras")
plt.show()


plt.figure()
df["genero"].value_counts().plot(kind="bar")
plt.title("Distribución por género")
plt.show()

plt.figure()
df["ciudad"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Distribución de ciudades")
plt.show()
```

## ¿Tendría sentido incluir un Análisis Exploratorio de Datos en un pipeline de datos?

**A mi punto de vista SI, es fundamental.
El EDA permite detectar errores, entender los datos y asegurar su calidad antes de procesarlos o usarlos en modelos y asi saber que lo que hacemos esta correcto
Sin EDA, un pipeline puede generar resultados incorrectos o poco confiables por lo cual lo mejor es siempre hacerlo**
